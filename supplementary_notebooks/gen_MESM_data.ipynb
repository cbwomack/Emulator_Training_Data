# Generate input data for MESM from SCM optimization

```python
# Compute migrated to scripts/supplementary_notebooks/gen_MESM_data.py (run
# standalone/scheduled; writes every scenario's CO2 emissions as a MESM-format
# .txt input file to data/MESM/emis_driven/MESM_inputs/).
mesm_inputs_dir = 'data/MESM/emis_driven/MESM_inputs'
scenarios_to_plot = ['H-ext', 'M', 'ML', 'L', 'VLLO-ext', 'VLHO', 'historical']
```

```python
# Sanity-check a subset of the written input files
for scen in scenarios_to_plot:
  co2_array = np.loadtxt(f'{mesm_inputs_dir}/{scen}_emis.txt', skiprows=2, usecols=(2,))
  plt.plot(co2_array, label=scen)
plt.ylim([-40, 100])
plt.legend()
```

```python
# Sanity-check the optimized (inverse) scenario's written input file
co2_array_opt = np.loadtxt(f'{mesm_inputs_dir}/opt_all_sine_emis.txt', skiprows=2, usecols=(2,))
plt.plot(co2_array_opt, label='all')
plt.legend()
```
