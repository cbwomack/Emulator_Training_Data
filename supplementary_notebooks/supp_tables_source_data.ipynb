## Table S2 (`tab:supp_opt_v_dyn`) — Multi row

Computed by the newly extended `scripts/6b_iteration0_ablation.py` (added a
`"Multi"` entry to `AGENT_CONFIG`/`FIG_GROUPS`): `Opt. Iter. 0` / `Opt. Final`
are read directly from the existing `checkpoints/multi_fig4_smooth/seed_sweep`
tier1-optimized checkpoints (same source as the Figure 4/5 Multi panels);
`Unopt. Rich` comes from 100 new `--mode control` runs (50 seeds x
{sine, gaussian} IC, `num_updates=0`) run on Engaging. All values are medians
across 50 seeds, evaluated on Priority 1.

**Flag**: every existing single-agent row uses Gaussian as the "best
performing" dynamically-rich IC (per the table's caption). For Multi,
**sine outperforms Gaussian** (lower median NRMSE) — the opposite of every
single-forcing agent. The row below uses sine, consistent with the caption's
"best performing" selection rule, but this means the caption's parenthetical
"(Gaussian)" is no longer accurate as a blanket statement once Multi is
added.

```python
import pickle
import numpy as np
import utils_inverse

N = 50
opt_iter0, opt_final = [], []
for seed in range(N):
    p = f'checkpoints/multi_fig4_smooth/seed_sweep/inverse_constant_tier1_multi_fig4_seed{seed}.pkl'
    ckpt = utils_inverse.load_inverse_ckpt(p)
    nrmse = utils_inverse.recover_nrmse_trajectory(ckpt)
    opt_iter0.append(float(nrmse[0]))
    opt_final.append(float(nrmse[-1]))

def load_control(ic):
    errs = []
    for seed in range(N):
        p = f'checkpoints/iteration0_controls/inverse_{ic}_tier1_multi_fig4_ic-{ic}_seed{seed}.pkl'
        with open(p, 'rb') as f:
            c = pickle.load(f)
        errs.append(float(c['errors'][0]))
    return errs

gauss = load_control('gaussian')
sine = load_control('sine')
best_ic = 'gaussian' if np.median(gauss) < np.median(sine) else 'sine'
best = gauss if best_ic == 'gaussian' else sine

df_table_s2_multi = pd.DataFrame([{
    "agent": "Multi",
    "Opt. Iter. 0": round(float(np.median(opt_iter0)), 3),
    "Unopt. Rich": round(float(np.median(best)), 3),
    "Opt. Final": round(float(np.median(opt_final)), 4),
    "Gain": f"{np.median(best) / np.median(opt_final):.1f}x",
    "best_ic_used": best_ic,
    "gaussian_median": round(float(np.median(gauss)), 3),
    "sine_median": round(float(np.median(sine)), 3),
}])
df_table_s2_multi
```
